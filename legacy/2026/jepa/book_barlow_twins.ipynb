```python
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

COLOR = "#948678"
# COLOR='#063336'


def make_correlation_matrix(
    N=20,
    noise_scale=0.5,
    total_size=3.0,       # inches (matches manim units 1:1 in the figure)
    seed=42,
    line_stroke_width=0.6,
    color=COLOR,
    fill_mode="alpha",    # "alpha": opacity = |M|; "blend": blend toward white (opaque fills)
    out_path="correlation_matrix.svg",
):
    """
    Matplotlib port of the ManimGL helper.
    Returns M (N x N symmetric) and writes an SVG of the shaded grid.
    """
    rng = np.random.default_rng(seed)
    M = np.eye(N) + noise_scale * rng.standard_normal((N, N))
    M = (M + M.T) / 2.0
    M = np.clip(M, -1.0, 1.0)

    cell = total_size / N
    half = total_size / 2.0

    fig = plt.figure(figsize=(total_size, total_size))
    ax = fig.add_axes([0, 0, 1, 1])   # fill the whole figure, no margins
    ax.set_xlim(-half, half)
    ax.set_ylim(-half, half)
    ax.set_aspect("equal")
    ax.axis("off")

    base = np.array(matplotlib.colors.to_rgb(color))

    # --- filled squares (drawn first so lines sit on top) ---
    for i in range(N):            # row (top -> bottom)
        for j in range(N):        # col (left -> right)
            x0 = -half + j * cell
            y0 = half - (i + 1) * cell
            v = float(abs(M[i, j]))
            if fill_mode == "alpha":
                fc = (*base, v)
            else:  # blend: shade of the base color, fully opaque
                fc = tuple(1.0 - (1.0 - base) * v)
            ax.add_patch(
                Rectangle((x0, y0), cell, cell, facecolor=fc, edgecolor="none",
                          linewidth=0, zorder=1)
            )

    # --- grid lines (N+1 verticals + N+1 horizontals) ---
    for k in range(N + 1):
        off = -half + k * cell
        ax.plot([off, off], [-half, half], color=color,
                linewidth=line_stroke_width, solid_capstyle="butt", zorder=2)
        ax.plot([-half, half], [off, off], color=color,
                linewidth=line_stroke_width, solid_capstyle="butt", zorder=2)

    fig.savefig(out_path, format="svg", transparent=True)
    plt.close(fig)
    return M
```

```python
M = make_correlation_matrix(N=20, out_path="data/ai_book_vol_2/4_jepa/graphics/bt_matrix_2.svg")
```
